# Conversational AI with Amazon Bedrock

This notebook demonstrates how to use Amazon Bedrock to create a conversational AI system. The main function `converse_marketplace_endpoint` is defined to interact with a Bedrock model endpoint.

Key components:
1. Function to interact with Bedrock endpoint
2. System prompt setting
3. List of diverse prompts for testing
4. Loop to iterate through prompts and get responses

The notebook uses an endpoint and includes a sleep timer between requests to avoid rate limiting.


```python
### Imports
import json
import boto3
import time
```

```python
# we get the bedrock runtime and the bedrock client
bedrock_runtime = boto3.client('bedrock-runtime')
bedrock = boto3.client('bedrock')
```

```python
def converse_marketplace_endpoint(id, user_prompt, system_prompt, max_tokens=2000, temperature=0):
    import traceback
    response = ''
    try:
        response = bedrock_runtime.converse(
            modelId=id,
            messages=[
                {
                    "role": "user",
                    "content": [
                        {
                            "text": user_prompt
                        }
                    ]
                }
            ],
            system=[ # Optional, remove this parameter if preferred
                { "text": system_prompt }
            ],
            inferenceConfig={ # Adjust/remove parameters as needed
                "temperature": temperature,
                "maxTokens": max_tokens,
                #"topP": top_p
            }
            #additionalModelRequestFields={ # Adjust/remove parameters as needed
            #}
        )
        # Extract and show the response text...
        print(response['output']['message']['content'][0]['text'])
        return
    except Exception as e:
        print(traceback.format_exc())
        return f"Error: {str(e)}"
```
